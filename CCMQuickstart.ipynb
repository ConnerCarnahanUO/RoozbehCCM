```python
import numpy as np
import DelayEmbedding.optimizedDErefactor2 as CCM
import os
```

```python
data_path = './DataSets/Logistic3-2/Logistic3TimeSeries.npy' #something to a timeseries in a numpy TxN format
TimeSeries = np.load(data_path) # Must be in the form [time step, var num]
outpath = f'./outputs/Logistic3-2/' #+ dataname
if not os.path.isdir(outpath):
    os.makedirs(outpath)
```

## CCM Parameters:

just run the cell if you don't care, but I have the meanings of everything here

```python
dim_min = 1 #Minimum Delay Dimension to test
dim_max = 10 #Maximum Delay Dimension to test and where the p-values and first iteration of CCM will be computed
kfolds = 5 #Cross Validation kfolds
delay = 1 #Delay length of the embedding in time steps
dt = 1 #For computing the lags so we don't have an insane number of lags computed/aren't computing lags on two short of interval
lag_min = -int(10/dt) #minimum tested lag
lag_max = int(10/dt) #maximum tested lag
neg_lags = np.arange(lag_min,0,int(1/dt))
pos_lags = np.arange(lag_max+1,step=int(1/dt))
lags = np.hstack([neg_lags,pos_lags])
lagstep = 1 # multiplies the lags if you want for a broader spread
random_projection = True #whether we should use random projection on the delay embeddings to make sure there is less cross correlation over the vector
compute_pvalue = True #whether we should compute the pvalue
n_surrogates = 50//kfolds #number of surrogate time series to generate for every kfold for p-value testing
normal_pval = False #Whether we want to assume a normal distribution for the surrogate null distribution (works to extend p-values and seems reasonable)
pval_threshold = 0.01 #The threshold on when we throw away connections that are insignificant for constructing the network graph
find_optimum_dims = True #Whether we want to do the actual dimension embedding parameter search
compute_noise_correlations=True #Whether we should do all of the noise correlation correction stuff for all of this
```

```python
CCM.DimensionParameterSearchLargeDatasetOpt2(TimeSeries,dim_min=dim_min, dim_max=dim_max, kfolds=kfolds, delay=delay,
                             lags=lags,lagstep=lagstep, random_projection = random_projection , mask = None, transform='none',
                             node_ratio = 1., compute_pvalue = True, n_surrogates = n_surrogates, normal_pval=normal_pval, pval_threshold=pval_threshold, 
                             min_pairs = 1, dim_search_stopping_num = 3, save=True, save_path = outpath, retain_test_set = True, 
                             max_processes = 100, early_stop = False, only_hubs = False, find_optimum_dims = find_optimum_dims, save_surrogates=False,
                             seed = 0,pval_correction='None',causal_peak_smoothing=1,bad_channels = None,noise_correlation_bins = 100,noise_correlation_shuffles = 100,compute_noise_correlations=compute_noise_correlations)
```
